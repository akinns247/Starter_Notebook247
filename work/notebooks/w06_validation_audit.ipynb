## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*
Use:

I checked the final feature list for direct ingredients of my proxy target and for identifier fields. The target was built from CTR, trend direction, and average position, while those direct target-related fields were excluded from the model features. The audit returned an empty set for direct target-related features, and neither client_id nor content_id was used as a feature. The leakage check therefore passed for this feature set.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Section 3: leakage audit for the final model feature set

model_features = [
    "search_volume",
    "competition",
    "cpc",
    "word_count",
    "char_count",
    "engagement_rate",
    "scroll_rate",
    "ai_traffic_pct",
    "impression_tier",
    "position_tier",
    "trend_pct"
]

target_related_features = [
    "ctr",
    "trend_direction",
    "avg_position"
]

identifier_features = [
    "client_id",
    "content_id"
]

direct_leakage = [
    feature for feature in model_features
    if feature in target_related_features
]

identifier_leakage = [
    feature for feature in model_features
    if feature in identifier_features
]

print("Model features:")
print(model_features)

print("\nDirect target-related features found in model:")
print(direct_leakage)

print("\nIdentifier features found in model:")
print(identifier_leakage)

print("\nLeakage audit result:")
if not direct_leakage and not identifier_leakage:
    print("PASS — no direct target-related or identifier features are present.")
else:
    print("REVIEW — possible leakage or identifier feature detected.")
```
